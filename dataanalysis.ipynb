```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Name" : ["A" , "B" , "C" , "D" , "E"],
    "Math" : [78,35,6,46,64],
    "Science" : [24,35,53,64,46],
    "English" : [12,3,53,53,64]
}
df = pd.DataFrame(data)
print(df)
```

```python
marks = df[["Maths", "Science", "English"]].values
average_marks = np.mean(marks, axis=1)
df["Average"] = average_marks
print(df)
```

```python
def grade(avg):
  if avg >= 90:
    return "A"
  elif avg >= 80:
    return "B"
  elif avg >= 70:
    return "C"
  elif avg >= 60:
    return "D"
  else:
    return "F"

df["Grade"] = df["Average"].apply(grade)
```

```python
plt.figure()
plt.bar(df["Name"], df["Average"], color='red')
plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.title("Average Marks of Students")
plt.show()
```

```python
#avg marks per subject
df[["Maths", "Science", "English"]].mean().plot(kind='bar', color=['red', 'black', 'yellow'])
plt.title("Average Marks per Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.show()
```
