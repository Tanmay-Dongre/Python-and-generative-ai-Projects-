```python
#impotring
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#load it data
df = pd.read_csv("iot_sensor_data.csv")
df["Timestamp"] = pd.to_datetime(df["Timestamp"])
print("IOT Sensor Data /n" , df)
```

```python
temp = df["Temperature"].values
mean_temp = np.mean(temp) # mean
std_temp = np.std(temp) #standard deviation
print("\n mean temprature = " , mean_temp)
print("Standard deviation is = " , std_temp)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(
    df["Timestamp"],
    df["Temperature"],
    marker="o"
)

plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("Temperature vs Time")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```
